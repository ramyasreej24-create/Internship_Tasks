## Q1. Frequency of each word in a sentence
**Pseudocode:** Make an empty dictionary. Split the sentence into words. For each word, add 1 to its count (start at 0 if unseen).

**Expected output (given example):** `{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}`

```python
# Q1. Frequency of each word in a sentence
def word_frequency(sentence):
    freq = {}
    for word in sentence.split():
        freq[word] = freq.get(word, 0) + 1
    return freq

# given example (first) + additional test (second)
print(word_frequency("python is easy and python is powerful"))
print(word_frequency("to be or not to be"))
```

## Q2. Frequency of each character in a word
**Pseudocode:** Make an empty dictionary. Go through each character; add 1 to its count. Insertion order keeps first-appearance order.

**Expected output (given example):** `{'b': 1, 'a': 3, 'n': 2}`

```python
# Q2. Frequency of each character in a word
def char_frequency(word):
    freq = {}
    for ch in word:
        freq[ch] = freq.get(ch, 0) + 1
    return freq

# given example (first) + additional test (second)
print(char_frequency("banana"))
print(char_frequency("hello"))
```

## Q3. First non-repeating character
**Pseudocode:** Count every character first. Then scan the word again left to right and return the first character whose count is 1; otherwise None.

**Expected output (given example):** `w`

```python
# Q3. First non-repeating character
def first_non_repeating(word):
    freq = char_frequency(word)
    for ch in word:
        if freq[ch] == 1:
            return ch
    return None

# given example (first) + additional test (second)
print(first_non_repeating("swiss"))
print(first_non_repeating("aabb"))
```

## Q4. Check whether two strings are anagrams
**Pseudocode:** Count characters of both strings. If the two count dictionaries are equal, they are anagrams.

**Expected output (given example):** `True`

```python
# Q4. Check whether two strings are anagrams
def are_anagrams(s1, s2):
    return char_frequency(s1) == char_frequency(s2)

# given example (first) + additional test (second)
print(are_anagrams("listen", "silent"))
print(are_anagrams("hello", "world"))
```

## Q5. First repeating character (earliest second occurrence)
**Pseudocode:** Keep a set of characters seen. Scan left to right; if a character is already in the set, return it. If the loop ends, return None.

**Expected output (given example):** `r`

```python
# Q5. First repeating character (earliest second occurrence)
def first_repeating(text):
    seen = set()
    for ch in text:
        if ch in seen:
            return ch
        seen.add(ch)
    return None

# given example (first) + additional test (second)
print(first_repeating("programming"))
print(first_repeating("abc"))
```

## Q6. Move all zeros to the end, keeping order of non-zeros
**Pseudocode:** Collect all non-zero items in order. Count how many zeros were dropped and append that many zeros.

**Expected output (given example):** `[3, 1, 2, 4, 0, 0, 0]`

```python
# Q6. Move all zeros to the end, keeping order of non-zeros
def move_zeros(nums):
    non_zeros = [x for x in nums if x != 0]
    return non_zeros + [0] * (len(nums) - len(non_zeros))

# given example (first) + additional test (second)
print(move_zeros([0,3,0,1,2,0,4]))
print(move_zeros([0,0,1]))
```

## Q7. Second highest distinct number
**Pseudocode:** Convert to a set to drop duplicates. If fewer than 2 values, return None. Remove the maximum, then the new maximum is the answer.

**Expected output (given example):** `30`

```python
# Q7. Second highest distinct number
def second_highest(nums):
    distinct = set(nums)
    if len(distinct) < 2:
        return None
    distinct.remove(max(distinct))
    return max(distinct)

# given example (first) + additional test (second)
print(second_highest([10,20,40,30,40,20]))
print(second_highest([5,5,5]))
```

## Q8. Remove duplicates preserving first-occurrence order
**Pseudocode:** Keep a 'seen' set and a result list. For each item not in seen, add it to both.

**Expected output (given example):** `[4, 2, 5, 7]`

```python
# Q8. Remove duplicates preserving first-occurrence order
def remove_duplicates(nums):
    seen = set()
    result = []
    for x in nums:
        if x not in seen:
            seen.add(x)
            result.append(x)
    return result

# given example (first) + additional test (second)
print(remove_duplicates([4,2,4,5,2,7,5]))
print(remove_duplicates([1,1,1]))
```

## Q9. Missing number from 1 to N
**Pseudocode:** Sum of 1..N is N(N+1)/2. Subtract the sum of the given list; the difference is the missing number.

**Expected output (given example):** `3`

```python
# Q9. Missing number from 1 to N
def missing_number(nums, n):
    return n * (n + 1) // 2 - sum(nums)

# given example (first) + additional test (second)
print(missing_number([1,2,4,5,6], 6))
print(missing_number([2,3,4,5], 5))
```

## Q10. First pair summing to target (second element found earliest)
**Pseudocode:** For each position j (left to right), check earlier positions i from the start. First time nums[i]+nums[j]==target, return [nums[i], nums[j]]. Else [].

**Expected output (given example):** `[2, 7]`

```python
# Q10. First pair summing to target (second element found earliest)
def two_sum_pair(nums, target):
    for j in range(len(nums)):
        for i in range(j):
            if nums[i] + nums[j] == target:
                return [nums[i], nums[j]]
    return []

# given example (first) + additional test (second)
print(two_sum_pair([2,7,11,15], 9))
print(two_sum_pair([1,2,3], 10))
```

## Q11. Number closest to zero (tie -> positive)
**Pseudocode:** Start with the first number as best. For each other number, replace best if its absolute value is smaller, or equal with a larger (positive) value.

**Expected output (given example):** `2`

```python
# Q11. Number closest to zero (tie -> positive)
def closest_to_zero(nums):
    best = nums[0]
    for x in nums[1:]:
        if abs(x) < abs(best) or (abs(x) == abs(best) and x > best):
            best = x
    return best

# given example (first) + additional test (second)
print(closest_to_zero([-8,-3,2,5,-2]))
print(closest_to_zero([-4,4,7]))
```

## Q12. Common elements, unique, order of first list
**Pseudocode:** Put list2 in a set. Walk list1; add an item to the result if it is in the set and not already in the result.

**Expected output (given example):** `[3, 4, 5]`

```python
# Q12. Common elements, unique, order of first list
def common_elements(list1, list2):
    in_second = set(list2)
    result = []
    for x in list1:
        if x in in_second and x not in result:
            result.append(x)
    return result

# given example (first) + additional test (second)
print(common_elements([1,2,3,4,5],[3,4,5,6,7]))
print(common_elements([1,2,2,3],[2,2,3]))
```

## Q13. Rotate list right by K positions
**Pseudocode:** Reduce K using K % length (handles K > length). Result = last K items + the rest. Handle empty list and K=0.

**Expected output (given example):** `[4, 5, 1, 2, 3]`

```python
# Q13. Rotate list right by K positions
def rotate_right(nums, k):
    if not nums:
        return []
    k %= len(nums)
    if k == 0:
        return nums[:]
    return nums[-k:] + nums[:-k]

# given example (first) + additional test (second)
print(rotate_right([1,2,3,4,5], 2))
print(rotate_right([1,2,3], 5))
```

## Q14. Maximum sum of two consecutive elements
**Pseudocode:** Start best as first two items' sum. Slide through the list adding neighbours and keep the maximum.

**Expected output (given example):** `14`

```python
# Q14. Maximum sum of two consecutive elements
def max_consecutive_sum(nums):
    best = nums[0] + nums[1]
    for i in range(1, len(nums) - 1):
        best = max(best, nums[i] + nums[i + 1])
    return best

# given example (first) + additional test (second)
print(max_consecutive_sum([4,7,2,9,5]))
print(max_consecutive_sum([-5,-1,-3]))
```

## Q15. All unique pairs summing to target
**Pseudocode:** Check every pair of different positions i<j. If sum equals target, store the sorted pair in a set (removes duplicates). Return the set sorted as lists.

**Expected output (given example):** `[[1, 8], [2, 7], [4, 5]]`

```python
# Q15. All unique pairs summing to target
def all_pairs(nums, target):
    pairs = set()
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pairs.add(tuple(sorted((nums[i], nums[j]))))
    return [list(p) for p in sorted(pairs)]

# given example (first) + additional test (second)
print(all_pairs([2,4,3,5,7,8,1], 9))
print(all_pairs([3,3,3], 6))
```
