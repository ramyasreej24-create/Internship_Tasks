```python
#Q1. Find the Shortest Word in a Sentence
n="Python makes coding easy"
words = n.split()
shortest = words[0]
for word in words:
    if len(word)<len(shortest):
      shortest=word
print(shortest)

```

```python
#Q2. Find the Largest Number in a List
lst=[12, 45, 7, 89, 23]
largest=lst[0]
for num in lst:
    if num > largest:
      largest=num
print(largest)
```

```python
#Q3. Find the smallest number in the list
lst=[12,45,7,89,23]
smallest=lst[0]
for num in lst:
  if num < smallest:
    smallest=num
print(smallest)
```

```python
#Q4. Find the First Longest Word When Ties Occur
sentence="Python makes coding useful"
words=sentence.split()
longest=words[0]
for word in words:
  if len(word) > len(longest):
    longest = words
print(longest)
```

```python
#Q5. Find the Word with the Maximum Number of Characters
text="Data engineering requires problem solving"
words = text.split()
Maximum = words[0]
for word in words:
  if len(words) > len(Maximum):
    Maximum = word
print(Maximum)
```

```python
#Q6.Count Vowels in a String
text = "Education"
vowels= "AEIOUaeiou"
count=0
for char  in text:
  if char in vowels:
    count += 1
print(count)

```

```python
# Q7. Count Consonants in a String
text="Python 123!"
consonants = "bcdfghjklmnpqrstvwxyz"
count=0
for char  in text.lower():
  if char in consonants:
    count += 1
print(count)

```

```python
#Q10. Find All Characters Appearing More Than Once
word = "programming"
t = []
for char in word:
    if word.count(char) > 1 and char not in t:
        t.append(char)
print(t)
```

```python
#Q11. Create a Dictionary of Words and Their First Characters
text = "python data engineering"
words = text.split()
r = {}
for word in words:
    r[word] = word[0]
print(r)
```

```python
#Q12. Create a Dictionary of Words and Their last Characters
text = "python data engineering"
words = text.split()
r = {}
for word in words:
    r[word] = word[-1]
print(r)
```

```python
# Q13. Create a Dictionary of Words and Their Uppercase Versions
n = "data science python"
words = n.split()
m = {}
for word in words:
    m[word] = word.upper()
print(m)
```

```python
#Q14. Create a Dictionary of Numbers and Their Squares
n = [2, 3, 4, 5]
o = {}
for m in n:
  o[m]=m**2
print(o)
```

```python
#Q15. creat a Dictionary of Numbers and Their Even/Odd Status
b = [1, 2, 3, 4, 5]
o ={}
for m in b:
    if m % 2 == 0:
        o[m] = "Even"
    else:
        o[m] = "Odd"
print(o)
```
